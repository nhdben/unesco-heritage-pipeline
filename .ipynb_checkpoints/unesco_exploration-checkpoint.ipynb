## 2.1 Pull Full Dataset

To get all 1,248 sites I loop through the API using two parameters:
- limit: how many records to return per request ( 100)
- offset: where to start from (moving forward by 100 each loop)

So the first request gets records 1-100, the second gets 101-200, and so on
until all sites are collected into one list.

```python
all_records = []
limit = 100
offset = 0
total = 1248

while offset < total:
    url = f"https://data.unesco.org/api/explore/v2.1/catalog/datasets/whc001/records?limit={limit}&offset={offset}"
    response = requests.get(url)
    batch = response.json().get("results", [])
    all_records.extend(batch)
    offset += limit
    print(f"fetched {len(all_records)} / {total}")

print("done")
```

## 4. Clean & Transform

Keeping only the columns needed, fixing data types, extracting
coordinates, and adding new columns to support the analysis.

```python
# keeping only useful columns
cols = [
    "name_en", "category", "date_inscribed", "danger",
    "region", "region_code", "states_names", "iso_codes",
    "transboundary", "area_hectares", "criteria_txt",
    "cultural_criteria", "natural_criteria", "coordinates"
]

df = df[cols].copy()
print(df.shape)
```

## 4.1 Fix Data Types

Converting danger and transboundary from strings to booleans, and
date_inscribed from a string to an integer so we can do math on it later.

```python
# fixing data types
df["danger"] = df["danger"].map({"True": True, "False": False})
df["transboundary"] = df["transboundary"].map({"True": True, "False": False})
df["date_inscribed"] = pd.to_numeric(df["date_inscribed"], errors="coerce")

print(df.dtypes)
```

## 4.2 Extract Coordinates

The coordinates column is a nested dictionary with lat and lon inside.
Extracting them into two separate columns so it can be used on Looker Studio
for the map visualization.

```python
# extract lat and lon from coordinates dictionary
df["latitude"] = df["coordinates"].apply(lambda x: x["lat"] if x else None)
df["longitude"] = df["coordinates"].apply(lambda x: x["lon"] if x else None)

df = df.drop(columns=["coordinates"])
print(df[["name_en", "latitude", "longitude"]].head())
```

## 4.3 Add New Columns

Adding three new columns to support the analysis:
- decade: groups the inscription year into decades (1970s, 1980s, etc.)
- is_arab_league: flags sites from Arab League member states
- criteria_count: counts how many inscription criteria a site meets

```python
# see all unique country names in the dataset
import itertools
all_countries = list(itertools.chain.from_iterable(df["states_names"]))
print(sorted(set(all_countries)))
```

```python
# group inscription year into decades
df["decade"] = (df["date_inscribed"] // 10 * 10).astype(str) + "s"

# flag arab league countries
arab_league = [
    "Algeria", "Bahrain", "Egypt", "Iraq",
    "Jordan", "Kuwait", "Lebanon", "Libya", "Mauritania", "Morocco",
    "Oman", "State of Palestine", "Qatar", "Saudi Arabia", "Sudan",
    "Syrian Arab Republic", "Tunisia", "United Arab Emirates", "Yemen"
]

df["is_arab_league"] = df["states_names"].apply(
    lambda x: any(country in arab_league for country in x) if x else False
)

# count inscription criteria
df["criteria_count"] = df["criteria_txt"].str.count(r"\(") 

print(df[["name_en", "decade", "is_arab_league", "criteria_count"]].head(10))
```
